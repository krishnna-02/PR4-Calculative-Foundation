PART A

Vector:
A vector is an ordered collection of numerical values. Here, each student's marks in four subjects are represented as one vector

```python
for i in range(5):
    print(df.iloc[i]['Student_ID'], X[i])
```

Calculate Norm-1 and Norm-2:
 Norm-1: Sum of the absolute values of vector elements
 Norm-2: Square root of the sum of squares of vector elements

```python
for i in range(5):
    norm1 = np.linalg.norm(X[i], ord=1)
    norm2 = np.linalg.norm(X[i], ord=2)

    print(df.iloc[i]['Student_ID'],
          "Norm-1:", norm1,
          "Norm-2:", round(norm2, 2))
```

Dot product measures the combined similarity of two vectors. The angle between vectors helps understand how similar their score patterns are
A=[75,72,87,81]
B=[53,56,66,72]

```python
A = X[0]
B = X[1]

dot_product = np.dot(A, B)

cos_theta = dot_product / (
    np.linalg.norm(A) * np.linalg.norm(B)
)

angle = np.degrees(np.arccos(np.clip(cos_theta, -1, 1)))

print("Dot Product:", dot_product)
print("Angle:", round(angle, 2), "degrees")
```

Cross product produces a vector perpendicular to two input vectors. It is defined for 3-dimensional vectors. 
A=[75,72,87]
B=[53,56,66]

```python
A = X.iloc[0, :3].to_numpy()
B = X.iloc[1, :3].to_numpy()
cross_product = np.cross(A, B)

print("Vector A:", A)
print("Vector B:", B)
print("Cross Product:", cross_product)
```

PART B

Form a Matrix of Students × Subjects:
A matrix is a rectangular arrangement of numbers in rows and columns. Here, rows represent students and columns represent subjects.

```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\KRISHNA\Downloads\student_performance_dataset_1000.csv")

A = df[['Maths', 'Science', 'English', 'Computer']].values

print("Student Matrix:")
print(A[:5])

print("Matrix Shape:", A.shape)
```

Matrix Addition
Matrix addition means adding the corresponding elements of two matrices. Both matrices must have the same dimensions.

```python
B = A.copy()

addition = A + B

print("Matrix Addition:")
print(addition[:5])
```

Matrix Multiplication
Matrix multiplication is performed by multiplying rows of the first matrix with columns of the second matrix.

```python
multiplication = np.dot(A.T, A)

print("Matrix Multiplication:")
print(multiplication)
```

Matrix Transpose
Transpose converts rows into columns and columns into rows.

```python
transpose = A.T

print("Transpose Matrix:")
print(transpose[:, :5])

print("Transpose Shape:", transpose.shape)
```

Matrix Inverse
The inverse of a matrix is a matrix that gives the identity matrix when multiplied by the original matrix.
Only square matrices can have an inverse, and the determinant must be non-zero.

```python
square_matrix = np.dot(A.T, A)

inverse = np.linalg.inv(square_matrix)

print("Inverse Matrix:")
print(np.round(inverse, 8))
```

Determinant
The determinant is a single numerical value calculated from a square matrix. It helps determine whether a matrix has an inverse.

```python
determinant = np.linalg.det(square_matrix)

print("Determinant:", determinant)
```

Compute the Eigenvalues and Eigenvectors of the Covariance Matrix
Eigenvalues: Show how much variation exists in each direction.
Eigenvectors: Show the directions in which the data varies.

Av=lambda v
A=Covariance matrix
v=Eigenvector
lambda=Eigenvalue

```python
X = df[["Maths", "Science", "English", "Computer"]]
cov_matrix = np.cov(X, rowvar=False)

print("Covariance Matrix:")
print(cov_matrix)
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```
